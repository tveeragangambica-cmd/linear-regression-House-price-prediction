```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("House Price Prediction Dataset.csv")
data
```

```python
weight = 0
bias = 0
x_data = data["Area"]
y_data = data["Price"]
```

```python
def predict(x):
    y_pred = weight*x + bias
    return y_pred
```

```python
def error(y_pred,y_data):
    error = y_pred - y_data
    return error
def loss(error):
    loss = np.mean(error**2)
    return loss    
```

```python
def optimize(x_data,y_data,error):
    epochs = range(1000)
    global weight,bias
    learning_rate = 0.000000001
    n = len(x_data)
    for epoch in epochs:
        pred = predict(x_data)
        errors = error(pred,y_data)
        d_weight = (2/n)*np.sum(errors*x_data)
        d_bias = (2/n)*np.sum(errors)
        weight = weight-learning_rate*d_weight
        bias = bias-learning_rate*d_bias
        l = loss(errors)
        print(f"Loss :{l},epoch:{epoch}")
optimize(x_data,y_data,error)        
```
